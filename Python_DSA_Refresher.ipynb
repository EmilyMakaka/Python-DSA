## Question 1
### Write a program that asks the user for a number of days.The program then prints out the number of seconds in the amount of days given.

```python
SECONDS_IN_A_DAY = 24 * 60 * 60  # Convert  days to seconds: 24 hours/day * 60 minutes/hour * 60 seconds/minute

# Get input from the user 
days = float(input("Enter the number of days: "))

# Multiply the number of days by the number of seconds 
seconds = days * SECONDS_IN_A_DAY

# Display the result to the user
print(f"{days} day(s) is equal to {seconds} seconds.")
```

## Question 2
### Write a program that asks a user to input the radius then the program calculates the volume of a sphere (the formula for the volume is (4/3)πr3). Use the exponential operator in python to compute (r3).

```python
import math

#Get the radius from the user
radius = float(input("Enter the radius of the sphere: "))

#Calculate the volume of the sphere
volume = (4/3) * math.pi * radius**3

#Display the result to the user
print(f"The volume of the sphere with radius {radius} is {volume} cubic units.")
```

## Question 3
### Using functions, write a program to compute the area and perimeter of a square. The program should ask the user to enter a number corresponding to the side length of the square and display the area and perimeter of the square.

```python
# Calculate area of a square
def calculate_area(side):
    return side **2 

# Calculate perimeter of a square
def calculate_perimeter(side):
    return 4 * side

# Ask the user for the length of the side of the square
side_length = float(input("Enter the length of the side of the square: "))

# Calculate  the area and perimeter
area = calculate_area(side_length)
perimeter = calculate_perimeter(side_length)

# Display the results to the user
print(f"The area of the square with side {side_length} is {area}.")
print(f"The perimeter of the square with side {side_length} is {perimeter}.")
```

## Question 4
### Write a program using functions that determines whether a character input by a user is uppercase or lower case.

```python
# Check whether the character is uppercase or lowercase
def check_case(character):
    if character.isupper():
        return f"'{character}' is an UPPERCASE letter."
    elif character.islower():
        return f"'{character}' is a lowercase letter."
    else:
        return f"'{character}' is not a letter, so it has no case."

# Ask the user to input a single character
user_char = input("Enter a character: ")

# Call the function and display the result
print(check_case(user_char))
```

## Question 5
### The following is pseudocode for a program being designed. Write the Python program equivalent of the same.
###### BEGIN
###### SET x TO 0, y TO 20
###### REPEAT
###### SUBTRACT 4 FROM y
###### ADD 2/y TO x
###### UNTIL
###### y IS LESS THAN 6
###### DISPLAY x
###### END


```python
x = 0
y = 20

while True:
    y = y - 4 # Subtract 4 from y
    x = x + (2 / y) # Add 2/y to x
    if y < 6: # Until y is less than 6 stop repeating once this is true
        break

# Display x
print(x)
```
